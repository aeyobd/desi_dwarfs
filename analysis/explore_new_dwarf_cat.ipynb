# Histograms and selection functions

```python
dwarfs_bgs_bright = dwarfs_new[dwarfs_new["IS_BGS_BRIGHT"]]
dwarfs_bgs_faint = dwarfs_new[dwarfs_new["IS_BGS_FAINT"]]
dwarfs_lowz = dwarfs_new[dwarfs_new["IS_LOWZ"]]
dwarfs_elg = dwarfs_new[dwarfs_new["IS_ELG"]]
```

```python
dwarfs_subsets = {
    "LOW Z": dwarfs_lowz,
    "BGS-Bright": dwarfs_bgs_bright,
    "BGS-Faint": dwarfs_bgs_faint,
    "ELG": dwarfs_elg,
}
```

```python
dwarfs_subsets_good = {
    "LOW Z": dwarfs_lowz[dwarfs_lowz["GOOD"]],
    "BGS-Bright": dwarfs_bgs_bright[dwarfs_bgs_bright["GOOD"]],
    "BGS-Faint": dwarfs_bgs_faint[dwarfs_bgs_faint["GOOD"]],
    "ELG": dwarfs_elg[dwarfs_elg["GOOD"]],
}
```

```python
fig, ax = plt.subplots()

labels = dwarfs_subsets.keys()
counts = [len(dwarfs_subsets[label]) / 1e3 for label in labels]
counts_useful = [np.sum(dwarfs_subsets[label]["GOOD"]) / 1e3 for label in labels]

ax.bar(labels, counts, color=arya.COLORS[:4], fill=None, lw=1, ec=arya.COLORS[:4], label="total")
ax.bar(labels, counts_useful, color=arya.COLORS[:4], label="useful")

ax.legend()
ax.set(
    xlabel = "sample",
    ylabel = "number of galaxies (in thousands)"
)
```

```python
fig, ax = plt.subplots()

labels = dwarfs_subsets.keys()
counts = [len(dwarfs_subsets[label]) / 1e3 for label in labels]
counts_quality = [np.sum((dwarfs_subsets[label]["DWARF_MASKBIT"] == 0) & (dwarfs_subsets[label]["MSTAR_MASKBIT"] == 0)) / 1e3 for label in labels]

ax.bar(labels, counts, color=arya.COLORS[:4], fill=None, lw=1, ec=arya.COLORS[:4], label="total")
ax.bar(labels, counts_quality, color=arya.COLORS[:4], label="unflaged")

ax.legend()
ax.set(
    xlabel = "sample",
    ylabel = "number of galaxies (in thousands)"
)
```

```python
fig, ax = plt.subplots()

labels = dwarfs_subsets.keys()
counts_useful_updated = [np.sum(dwarfs_subsets[label]["GOOD"] & dwarfs_subsets[label]["PHOTOMETRY_UPDATED"] ) / 1e3 for label in labels]

ax.bar(labels, counts_useful, color=arya.COLORS[:4], fill=None, lw=1, ec=arya.COLORS[:4], label="useful")
ax.bar(labels, counts_useful_updated, color=arya.COLORS[:4], label="updated phot")

ax.legend()
ax.set(
    xlabel = "sample",
    ylabel = "number of galaxies (in thousands)"
)
```

```python
fig, ax = plt.subplots()

labels = dwarfs_subsets.keys()
counts_useful_updated = [np.sum(dwarfs_subsets[label]["GOOD"] & (dwarfs_subsets[label]["DWARF_MASKBIT"] == 0) & (dwarfs_subsets[label]["MSTAR_MASKBIT"] == 0) ) / 1e3 for label in labels]

ax.bar(labels, counts_useful, color=arya.COLORS[:4], fill=None, lw=1, ec=arya.COLORS[:4], label="useful")
ax.bar(labels, counts_useful_updated, color=arya.COLORS[:4], label="unflaged")

ax.legend()
ax.set(
    xlabel = "sample",
    ylabel = "number of galaxies (in thousands)"
)
```

```python
fig, axs = plt.subplots(1, 2, figsize=(6, 3))

plt.sca(axs[0])
plt.scatter(dwarfs_new["RA_TARGET"], d_old, s=1, lw=0, alpha=0.3)
plt.xlabel("RA")
plt.ylabel("$D_L$ / MPC")
plt.xlim(0, 360)
plt.ylim(0, 50)
plt.title("Redshift distance")

plt.sca(axs[1])
plt.scatter(dwarfs_new["RA_TARGET"], d_new, s=1, lw=0, alpha=0.3,)
plt.xlabel("RA")
plt.ylabel("$D_L$ / MPC")
plt.xlim(0, 360)
plt.ylim(0, 50)

plt.title("NAM distance")

plt.savefig("figures/d_ra_orig_nam.pdf")
```

```python
def dist_mod(d_Mpc):
    return 5 * np.log10(d_Mpc * 1e5)
```

```python
fig, ax = plt.subplots()

density_scatter(d_old, d_new, xlim=(0, 50), ylim=(0, 50), bins=300)
plt.xlabel("original (CMB) distance / Mpc")
plt.ylabel("NAM distance / Mpc")

ax.set(
    aspect=1
)

plt.savefig("figures/z_orig_nam.pdf")
```

```python
dist_mod(50)
```

```python
fig, ax = plt.subplots()

density_scatter(dist_mod(d_old), dist_mod(d_new), xlim=(28, 34), ylim=(28, 34), bins=300)
plt.xlabel("original (CMB) DM (mag)")
plt.ylabel("NAM DM (mag)")

ax.set(
    aspect=1
)
plt.plot([28, 34], [28, 34])

plt.savefig("figures/dm_orig_nam.pdf")
```
