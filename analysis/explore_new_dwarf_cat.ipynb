```python
import matplotlib.pyplot as plt
import numpy as np
# import arya
# arya.use("mnras")
```

```python
from astropy.table import Table, hstack
```

# Data loading

```python
dwarf_main = Table.read("../data/desi_dr1_dwarf_catalog.fits", hdu=1)
dwarf_fsf = Table.read("../data/desi_dr1_dwarf_catalog.fits", hdu=4)
```

```python
dwarfs_new = hstack([dwarf_main, dwarf_fsf])
```

# Histograms and selection functions

```python
dwarfs_bgs_bright = dwarfs_new[dwarfs_new["IS_BGS_BRIGHT"]]
dwarfs_bgs_faint = dwarfs_new[dwarfs_new["IS_BGS_FAINT"]]
dwarfs_lowz = dwarfs_new[dwarfs_new["IS_LOWZ"]]
dwarfs_elg = dwarfs_new[dwarfs_new["IS_ELG"]]
```

```python
fig, ax = plt.subplots()

ax.hist(dwarfs_bgs_bright["MAG_R"], histtype="step", label="BGS-Bright")
ax.hist(dwarfs_bgs_faint["MAG_R"], histtype="step", label="BGS-Faint")
ax.hist(dwarfs_lowz["MAG_R"], histtype="step", label="LOW-Z")
ax.hist(dwarfs_elg["MAG_R"], histtype="step", label="ELG")

ax.legend()
ax.set(
    xlabel = "r mag",
    ylabel = "count"
)
```

```python
fig, ax = plt.subplots()

ax.hist(dwarfs_bgs_bright["Z"], histtype="step", label="BGS-Bright")
ax.hist(dwarfs_bgs_faint["Z"], histtype="step", label="BGS-Faint")
ax.hist(dwarfs_lowz["Z"], histtype="step", label="LOW-Z")
ax.hist(dwarfs_elg["Z"], histtype="step", label="ELG")

ax.legend()
ax.set(
    xlabel = "r mag",
    ylabel = "count"
)
```
