The main query is below. We simply join on target id to select only the galaxies which are dwarfs

```python
query =\
"""
SELECT 
    *
FROM
    desi_dr1.stellar_mass_emline e
JOIN
    desi_dr1.photometry p
ON
    e.TARGETID = p.TARGETID
WHERE
    MASS_CG < 1e9
"""
```

```python

R = qc.query(sql=query,fmt='table', timeout=30000)
```

```python
R.write("dwarfs_cigale.fits", overwrite=True)
```
