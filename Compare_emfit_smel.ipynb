How well do flux measurements between both cataloguees match?

```python
import arya
```

```python
from astropy.table import join, Table

import numpy as np
import matplotlib.pyplot as plt
```

```python
emfit = Table.read("emfit_dwarfs.fits")
```

```python
emfit
```

```python
smel = Table.read("stellar_mass_emline_dwarfs.fits")
```

```python
joined = join(emfit, smel, "targetid", table_names=["emfit", "smel"])
```

```python
joined
```

```python
for c in joined.colnames:
    print(c)
```

```python
def plot_line_ratio(linename):
    x = joined[f"{linename}_flux_emfit"]
    y = joined[f"{linename}_flux_smel"]
    fig, ax = plt.subplots()
    
    plt.scatter(np.log10(x), np.log10(y/x), 
                s=1, lw=0, color="k", alpha=0.01)

    plt.axhline(0)

    ax.set(
        ylim = (-0.5, 0.5), 
        xlim = (0, 3),
        xlabel = f"log {linename} flux emfit",
        ylabel = f"log flux smel / emfit"
    )
```

```python
plt.rcParams.clear()
```

```python
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["text.usetex"] = False
plt.rcParams["font.family"] = "serif"
```

```python
def plot_xy(xname, yname):
    x = joined[xname]
    y = joined[yname]
    fig, ax = plt.subplots()
    
    plt.scatter(np.log10(x), np.log10(y/x), 
                s=1, lw=0, color="k", alpha=0.01)

    plt.axhline(0)

    ax.set(
        ylim = (-0.5, 0.5), 
        xlim = (0, 3),
        xlabel = f"log {xname} flux emfit",
        ylabel = f"log flux smel / emfit"
    )
```
